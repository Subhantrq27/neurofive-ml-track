## 6. Missing values

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_summary = pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
missing_summary[missing_summary["missing_count"] > 0].sort_values("missing_count", ascending=False)
```

## 7. Categorical vs. numerical columns

```python
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)
```

## 8. Data story — what we found

- The dataset has **891 rows and 12 columns**, one row per passenger.
- Three columns have missing values: **`Cabin`** is missing for the vast majority of
  passengers (~77%), **`Age`** is missing for about **1 in 5** passengers (~20%),
  and **`Embarked`** is missing just **2** values — small enough to drop or fill safely.
- The columns split into **7 numerical** (`PassengerId`, `Survived`, `Pclass`, `Age`,
  `SibSp`, `Parch`, `Fare`) and **5 categorical** (`Name`, `Sex`, `Ticket`, `Cabin`,
  `Embarked`). Note `Survived` and `Pclass` are numeric but really *encoded categories*.
- `Age` and `Fare` are right-skewed (a handful of expensive tickets and older
  passengers pull the mean above the median), and `Fare` has a minimum of 0,
  which is worth investigating later (free tickets? data entry issue?).
- Overall survival rate is around 38%, and the dataset is a solid, well-known
  starting point for classification — but `Age` and `Cabin` will need a clear
  missing-data strategy before modeling.
